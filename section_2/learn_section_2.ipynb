While loop

```python
i = 0
while i < 5:
    print(f"value of i is: {i}")
    i+=1
```

```python
i = 3
while True:
    print(f"value of i:{i}")
    if i >= 5:
        print(f"value of i: {i} >= 5, breaking out of while loop")
        break
    i += 1
```

```python
min_length = 5

while True:
    name = input("Enter your name: ")
    if len(name) >= min_length and name.isprintable() and name.isalpha():
        print("Conditions met, breaking")
        break
    else:
        print("Conditions have not met, let's retry again.")

print(f"Hello, {name}")
```

```python
a = 0

while a < 10:
    a += 1
    if a % 2 == 0:
        continue
    print(a)
```

```python
# l = [1,10,2,3]
l = [1,2,4]
val = 10

idx = 0

while idx < len(l):
    if l[idx] == val:
        print(f"found {val} on index {idx}")
        break
    idx += 1
else:
    print(f"Couldn't find val: {val} in l: {l}, appending val")
    l.append(val)
    print(l)
```

Break, Continue and Try statement

```python
a = 10
b = 0

try:
    print(a/b)
except ZeroDivisionError as e:
    print(f"error : {e}")
finally:
    print("This always executes")
```

```python
a = 1
b = 2

while a < 4:
    a += 1
    b -= 1
    print("------------------------------------")
    try:
        result = a/b
        print(f"main loop, a: {a} / b: {b} = {result}")
    except ZeroDivisionError as e:
        print(f"except block, a: {a} / b: {b} = {e}")
        print("will continue")
        continue
    finally:
        print(f"finally block, a: {a} / b: {b}")
        print("$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$")
else:
    print("runs if no break occured in while loop")
```
