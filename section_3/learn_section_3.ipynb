Garbage collection

```python
import gc
import ctypes

def get_ref_count(address):
    return ctypes.c_long.from_address(address).value
```

```python
def object_by_id(object_id):
    for obj in gc.get_objects():
        if id(obj) == object_id:
            return "Object exists"
    return "Not found"
```

```python
class A:
    def __init__(self) -> None:
        self.b =  B(self)
        print(f"A: self: {hex(id(self))}, b: {hex(id(self.b))}")

class B:
    def __init__(self, a) -> None:
        self.a = a
        print(f"B: self: {hex(id(self))}, a: {hex(id(self.a))}")
```

```python
gc.disable()
```

```python
my_var = A()
```

```python
hex(id(my_var))
```

```python
print(f"b: {hex(id(my_var.b))}")
print(f"a: {hex(id(my_var.b.a))}")
```

```python
a_id  = id(my_var)
b_id = id(my_var.b)
```

```python
print(f"a_id: {hex(a_id)}")
print(f"b_id: {hex(b_id)}")
```

```python
get_ref_count(a_id)
```

```python
get_ref_count(b_id)
```

```python
object_by_id(a_id)
```

```python
object_by_id(b_id)
```

```python
my_var = None
```

```python
get_ref_count(a_id)
```

```python
get_ref_count(b_id)
```

Function arguments and mutability

```python
def process(s):
    print(f"Initial s # = {id(s)}")
    s = s + 'world'
    print(f"Final s# = {id(s)}")
```

```python
my_var = "hello"
print(f"my_var # = {id(my_var)}")
```

```python
process(my_var)
```

```python
id(my_var)
```

```python
def  modify_list(lst):
    print(f"Initial lst # = {id(lst)}")
    lst.append(100)
    print(f"Final lst # = {id(lst)}")
```

```python
my_list = [1,2,3]
id(my_list)
```

```python
modify_list(my_list)
```

```python
my_list
```

```python
def modify_tuple(t):
    print(f"Initial t # = {id(t)}")
    t[0].append(100)
    print(f"Final t# = {id(t)}")
```

```python
my_tuple = ([12,33],1, 2)
```

```python
id(my_tuple)
```

```python
modify_tuple(my_tuple)
```

```python
my_tuple
```

```python
def cube(a):
    return a ** 3
```

```python
def select_function(fn_id):
    if fn_id == 1:
        return square
    else:
        return cube
```

```python
type(f)
```

```python
f is square
```

```python
f(2)
```

```python
select_function(2)(3)
```

```python
def exec_function(fn, n):
    return fn(n)
```

```python
exec_function(cube, 3)
```
