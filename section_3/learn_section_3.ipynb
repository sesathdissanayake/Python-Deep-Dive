Garbage collection

```python
import gc
import ctypes

def get_ref_count(address):
    return ctypes.c_long.from_address(address).value
```

```python
def object_by_id(object_id):
    for obj in gc.get_objects():
        if id(obj) == object_id:
            return "Object exists"
    return "Not found"
```

```python
class A:
    def __init__(self) -> None:
        self.b =  B(self)
        print(f"A: self: {hex(id(self))}, b: {hex(id(self.b))}")

class B:
    def __init__(self, a) -> None:
        self.a = a
        print(f"B: self: {hex(id(self))}, a: {hex(id(self.a))}")
```

```python
gc.disable()
```

```python
my_var = A()
```

```python
hex(id(my_var))
```

```python
print(f"b: {hex(id(my_var.b))}")
print(f"a: {hex(id(my_var.b.a))}")
```

```python
a_id  = id(my_var)
b_id = id(my_var.b)
```

```python
print(f"a_id: {hex(a_id)}")
print(f"b_id: {hex(b_id)}")
```

```python
get_ref_count(a_id)
```

```python
get_ref_count(b_id)
```

```python
object_by_id(a_id)
```

```python
object_by_id(b_id)
```

```python
my_var = None
```

```python
get_ref_count(a_id)
```

```python
get_ref_count(b_id)
```

```python
object_by_id(a_id)
```

```python
object_by_id(b_id)
```

```python
gc.collect()
```
